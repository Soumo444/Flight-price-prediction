## EDA With Red Wine Data

Data Set Information:

The two datasets are related to red and white variants of the Portuguese "Vinho Verde" wine.  Due to privacy and logistic issues, only physicochemical (inputs) and sensory (the output) variables are available (e.g. there is no data about grape types, wine brand, wine selling price, etc.).

These datasets can be viewed as classification or regression tasks. The classes are ordered and not balanced (e.g. there are many more normal wines than excellent or poor ones). Outlier detection algorithms could be used to detect the few excellent or poor wines. Also, we are not sure if all input variables are relevant. So it could be interesting to test feature selection methods.


Attribute Information:

Input variables (based on physicochemical tests):
- 1 - fixed acidity
- 2 - volatile acidity
- 3 - citric acid
- 4 - residual sugar
- 5 - chlorides
- 6 - free sulfur dioxide
- 7 - total sulfur dioxide
- 8 - density
- 9 - pH
- 10 - sulphates
- 11 - alcohol

Output variable (based on sensory data):
- 12 - quality (score between 0 and 10)

```python
import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"

df = pd.read_csv(url, sep=";")

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the figure size
plt.figure(figsize=(10, 6))

# Create a heatmap of the correlation matrix
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

# Add a title
plt.title('Feature Correlation Heatmap', fontsize=14)

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the figure size
plt.figure(figsize=(8, 5))

# Create a count plot for the 'quality' column
sns.countplot(x='quality', data=df, palette='Blues_d')

# Add labels and a title
plt.xlabel("Wine Quality")
plt.ylabel("Count")
plt.title("Wine Quality Distribution")

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Loop through every column in the dataframe and plot a histogram with KDE
for column in df.columns:
    plt.figure(figsize=(8, 5))
    sns.histplot(df[column], kde=True)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the figure size
plt.figure(figsize=(8, 6))

# Create a boxplot for alcohol versus quality
sns.boxplot(x='quality', y='alcohol', data=df)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the figure size
plt.figure(figsize=(9, 6))

# Create a colorful scatter plot using 'hue' to separate quality and a bright palette
sns.scatterplot(x='alcohol', y='pH', hue='quality', palette='Set1', data=df, s=60, alpha=0.8)

# Make it easier to read with a grid and proper title
plt.title('Alcohol vs pH Colored by Wine Quality', fontsize=14, fontweight='bold')
plt.xlabel('Alcohol', fontsize=12)
plt.ylabel('pH', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Move the legend outside or clean it up nicely
plt.legend(title='Quality', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
